# 🧱 BLOCO 3 – Personagem e animações
**Objetivo:** o personagem anda **uma casa por vez** com as setas, com animação de passo (pulinho) e de batida (tonto, com estrelinhas).

```python
LINHA_LIMITE = 300

TEMPO_PASSO = 12
TEMPO_BATIDA = 60

personagem = {"x": 250, "y": 550, "animacao": "parado", "tempo": 0, "faixa": 0}
```

| Linha | O que faz |
|---|---|
| `LINHA_LIMITE = 300` | O personagem não sobe acima do meio da tela. Dali para cima, quem desce é o mundo (Bloco 5) |
| `TEMPO_PASSO = 12` | O passo dura 12 quadros (0,2 s) |
| `TEMPO_BATIDA = 60` | A batida dura 60 quadros (1 s) |
| `"x": 250, "y": 550` | Posição **na tela** (canto superior esquerdo da imagem). Começa no meio, lá embaixo |
| `"animacao"` | `"parado"`, `"andando"` ou `"batida"` |
| `"tempo"` | Quantos quadros faltam para a animação acabar |
| `"faixa"` | Quantas faixas ele está **acima da largada** no mapa (diferente do `y`, que é na tela) |

```python
def comecar_animacao(personagem, nome, duracao):
    personagem["animacao"] = nome
    personagem["tempo"] = duracao

def atualizar_animacao(personagem):
    if personagem["tempo"] > 0:
        personagem["tempo"] = personagem["tempo"] - 1
    if personagem["animacao"] == "andando" and personagem["tempo"] == 0:
        personagem["animacao"] = "parado"

def terminou_batida(personagem):
    return personagem["animacao"] == "batida" and personagem["tempo"] == 0
```

**As animações funcionam como um cronômetro.**
| Linha | O que faz |
|---|---|
| `comecar_animacao` | Liga uma animação: guarda o nome e quantos quadros ela dura |
| `if tempo > 0: tempo = tempo - 1` | Chamada todo quadro: desconta 1 (sem ficar negativo) |
| `if animacao == "andando" and tempo == 0` | O passo acabou: volta a ficar `"parado"` |
| `terminou_batida` | Devolve `True` quando a batida acabou (o loop usa isso para dar game over) |

💡 A função **muda o dicionário original**: por isso `comecar_animacao` e `atualizar_animacao` não precisam de `return`.

```python
def mover_personagem(personagem, tecla):
    if personagem["animacao"] == "batida":
        return False
    x_antes = personagem["x"]
    faixa_antes = personagem["faixa"]
    precisa_rolar = False
    if tecla == pygame.K_UP:
        personagem["faixa"] = personagem["faixa"] + 1
        if personagem["y"] > LINHA_LIMITE:
            personagem["y"] = personagem["y"] - CASA
        else:
            precisa_rolar = True
    if tecla == pygame.K_DOWN and personagem["y"] < ALTURA - CASA:
        personagem["y"] = personagem["y"] + CASA
        personagem["faixa"] = personagem["faixa"] - 1
    if tecla == pygame.K_LEFT and personagem["x"] > 0:
        personagem["x"] = personagem["x"] - CASA
    if tecla == pygame.K_RIGHT and personagem["x"] < LARGURA - CASA:
        personagem["x"] = personagem["x"] + CASA
    if personagem["x"] != x_antes or personagem["faixa"] != faixa_antes:
        comecar_animacao(personagem, "andando", TEMPO_PASSO)
    return precisa_rolar
```

**Anda uma casa e avisa se o mundo precisa descer.**
| Linha | O que faz |
|---|---|
| `if animacao == "batida": return False` | Tonto depois da batida: sai da função sem andar |
| `x_antes` / `faixa_antes` | Guarda onde estava, para saber depois se andou |
| `K_UP` → `faixa + 1` | Para cima ele sempre avança uma faixa no mapa... |
| `if y > LINHA_LIMITE: y = y - CASA` | ...e, se ainda não chegou no meio da tela, sobe **na tela** (y diminui = sobe) |
| `else: precisa_rolar = True` | Já está no meio: não sobe; avisa que **o mundo** tem que descer |
| `K_DOWN` | Desce na tela e volta uma faixa. Não passa da última linha (`ALTURA - CASA` = 550) |
| `K_LEFT` / `K_RIGHT` | Muda o x, sem sair da tela (`LARGURA - CASA` = 550) |
| `if x != x_antes or faixa != faixa_antes` | Andou? Então começa a animação do passo |
| `return precisa_rolar` | Devolve True/False para o loop chamar `rolar_mundo` |

💡 **Por que `ALTURA - CASA` e não `ALTURA`?** x e y são o canto de cima da imagem: em y = 600 ela já estaria fora da tela.

```python
def imagem_do_personagem(personagem):
    if personagem["animacao"] == "andando":
        quadro = (personagem["tempo"] // 6) % 2
        return IMAGENS["andando"][quadro]
    if personagem["animacao"] == "batida":
        quadro = (personagem["tempo"] // 8) % 2
        return IMAGENS["batida"][quadro]
    return IMAGENS["parado"]
```

**Escolhe qual imagem mostrar agora.**
| Linha | O que faz |
|---|---|
| `tempo // 6` | Divisão inteira: muda de valor a cada 6 quadros |
| `% 2` | Resto da divisão por 2: fica alternando **0, 1, 0, 1...** |
| `IMAGENS["andando"][quadro]` | Pega o quadro 0 ou 1 da lista |
| `return IMAGENS["parado"]` | Se não está andando nem batendo |

💡 **`(tempo // 6) % 2`** é o truque da animação: as duas imagens se alternam a cada 6 quadros.

```python
def desenhar_personagem(tela, personagem):
    x = personagem["x"]
    y = personagem["y"]
    if personagem["animacao"] == "andando":
        y = y - min(personagem["tempo"], TEMPO_PASSO - personagem["tempo"])
    if personagem["animacao"] == "batida" and personagem["tempo"] > 0:
        x = x + random.randint(-3, 3)
    tela.blit(imagem_do_personagem(personagem), (x, y))

def voltar_ao_comeco(personagem):
    personagem["x"] = 250
    personagem["y"] = 550
    personagem["animacao"] = "parado"
    personagem["tempo"] = 0
    personagem["faixa"] = 0
```

| Linha | O que faz |
|---|---|
| `min(tempo, TEMPO_PASSO - tempo)` | **Pulinho:** vai de 0 até 6 e volta a 0 durante o passo. Subtrair do y = subir |
| `random.randint(-3, 3)` | **Tremendo:** na batida, mexe o x um pouco a cada quadro |
| `tela.blit(imagem, (x, y))` | **Cola** a imagem na tela, na posição (x, y) |
| `voltar_ao_comeco` | Põe o personagem de volta na largada, parado (usado ao começar uma partida) |

💡 O pulinho e o tremor mudam só **onde desenha**: `personagem["x"]` e `["y"]` não mudam.

# 🧱 BLOCO 4 – Obstáculos
**Objetivo:** pessoas que andam sozinhas pelas faixas, de um lado para o outro. Cada obstáculo é **um dicionário**, e todos ficam numa **lista**.

```python
TIPOS_OBSTACULO = {
    "aluno":      {"imagens": IMAGENS["alunos"],         "largura": CASA,     "velocidades": [1, 2, 3], "quantos": 2},
    "professora": {"imagens": [IMAGENS["professora"]],   "largura": CASA,     "velocidades": [1, 2],    "quantos": 2},
    "diretor":    {"imagens": [IMAGENS["diretor"]],      "largura": CASA,     "velocidades": [1],       "quantos": 2},
    "skate":      {"imagens": [IMAGENS["skate"]],        "largura": CASA,     "velocidades": [4, 5],    "quantos": 2},
    "zelador":    {"imagens": [IMAGENS["carrinho"]],     "largura": 2 * CASA, "velocidades": [3, 4, 5], "quantos": 1},
    "merendeira": {"imagens": [IMAGENS["merendeira"]],   "largura": 2 * CASA, "velocidades": [2, 3],    "quantos": 1},
    "bicicleta":  {"imagens": [IMAGENS["bicicleta"]],    "largura": 2 * CASA, "velocidades": [5, 6],    "quantos": 1},
}

obstaculos = []
```

**Os 7 tipos de obstáculo são dados**, não `if`s. É um **dicionário de dicionários**.
| Chave | O que é |
|---|---|
| `"imagens"` | Imagens possíveis (o aluno tem 3 roupas; os outros, 1) |
| `"largura"` | `CASA` = 1 casa; `2 * CASA` = 2 casas (quem empurra carrinho ou anda de bicicleta) |
| `"velocidades"` | Pixels por quadro que podem ser sorteados: diretor bem lento, bicicleta bem rápida |
| `"quantos"` | Quantos aparecem juntos numa faixa |
| `obstaculos = []` | A lista dos obstáculos na tela. Começa vazia: eles são sorteados faixa por faixa (Bloco 5) |

Ex: `TIPOS_OBSTACULO["skate"]["velocidades"]` → `[4, 5]`.

```python
def criar_obstaculo(x, y, velocidade, largura, imagem):
    if velocidade < 0:
        imagem = pygame.transform.flip(imagem, True, False)
    return {"x": x, "y": y, "velocidade": velocidade, "largura": largura, "imagem": imagem}
```

**Fábrica de obstáculos:** recebe os valores e **devolve** um dicionário pronto.
| Linha | O que faz |
|---|---|
| `if velocidade < 0:` | Anda para a esquerda? (positivo → direita, negativo → esquerda) |
| `pygame.transform.flip(imagem, True, False)` | **Espelha** a imagem na horizontal (não na vertical), para olhar para onde anda |
| `return {...}` | Monta o dicionário do obstáculo e devolve |

```python
def mover_obstaculos(obstaculos):
    for obstaculo in obstaculos:
        obstaculo["x"] = obstaculo["x"] + obstaculo["velocidade"]
        if obstaculo["x"] > LARGURA:
            obstaculo["x"] = -obstaculo["largura"]
        if obstaculo["x"] < -obstaculo["largura"]:
            obstaculo["x"] = LARGURA

def desenhar_obstaculos(tela, obstaculos):
    for obstaculo in obstaculos:
        tela.blit(obstaculo["imagem"], (obstaculo["x"], obstaculo["y"]))
```

| Linha | O que faz |
|---|---|
| `for obstaculo in obstaculos:` | Passa por cada dicionário da lista |
| `x = x + velocidade` | Anda. Velocidade negativa → x diminui → vai para a esquerda |
| `if x > LARGURA: x = -largura` | Saiu pela **direita**: reaparece escondido logo antes da borda **esquerda** |
| `if x < -largura: x = LARGURA` | Saiu inteiro pela **esquerda**: reaparece na borda **direita** |
| `tela.blit(obstaculo["imagem"], ...)` | Desenha a imagem de cada um na posição dele |

💡 Usamos `-largura` para ele entrar na tela **aos poucos**, em vez de "piscar" já dentro dela.

# 🧱 BLOCO 5 – Mapa infinito
**Objetivo:** a escola não acaba nunca.

🔁 **A ideia:** o personagem não sobe além do meio da tela. Quando ele aperta ↑ lá, quem se mexe é o **mundo**: tudo desce uma faixa, nasce uma faixa nova em cima e a de baixo some. Parece que a câmera segue o personagem.

🗺️ **Os 3 mapas** se repetem nesta ordem: corredor das salas → cantina → parquinho → corredor...
| Mapa | Faixa `"livre"` (segura) | Faixa `"corredor"` (tem obstáculos) |
|---|---|---|
| Corredor das salas | Piso de granilite | Corredor de madeira |
| Cantina | Piso xadrez | Passagem de lajota |
| Parquinho | Grama | Trilha de terra |

```python
FAIXAS_POR_FUNDO = 12

FAIXAS_FUNDO = [
    ["livre", "corredor", "corredor", "livre", "corredor", "corredor",
     "corredor", "livre", "corredor", "corredor", "livre", "livre"],
    ["livre", "corredor", "corredor", "livre", "corredor", "corredor",
     "livre", "corredor", "corredor", "corredor", "livre", "livre"],
    ["livre", "corredor", "livre", "corredor", "corredor", "corredor",
     "livre", "corredor", "corredor", "livre", "corredor", "livre"],
]

OBSTACULOS_DO_MAPA = [
    ["aluno", "aluno", "professora", "diretor", "zelador"],
    ["aluno", "aluno", "merendeira", "zelador"],
    ["aluno", "aluno", "skate", "bicicleta"],
]

mundo = {"linhas_roladas": 0}
```

| Linha | O que faz |
|---|---|
| `FAIXAS_POR_FUNDO = 12` | Cada fundo tem 12 faixas (600 / 50) |
| `FAIXAS_FUNDO` | **Lista de listas:** para cada mapa, o tipo de cada faixa **de cima para baixo**. Tem que bater com o desenho da imagem |
| `OBSTACULOS_DO_MAPA` | Para cada mapa, quem pode aparecer nele (merendeira só na cantina, skate só no parquinho...) |
| `"aluno", "aluno"` | Repetido de propósito: o aluno tem o dobro de chance no sorteio |
| `mundo = {"linhas_roladas": 0}` | Quantas faixas o mundo já desceu desde a largada |

```python
def qual_fundo(numero):
    return (numero // FAIXAS_POR_FUNDO) % len(FUNDOS)

def tipo_da_faixa(numero):
    linha = FAIXAS_POR_FUNDO - 1 - numero % FAIXAS_POR_FUNDO
    return FAIXAS_FUNDO[qual_fundo(numero)][linha]
```

**As faixas do mapa são numeradas de baixo para cima:** 0 = largada, 1, 2, 3...
| Linha | O que faz |
|---|---|
| `numero // 12` | Em qual "cópia" de fundo a faixa está: 0 (faixas 0 a 11), 1 (12 a 23), 2... |
| `% len(FUNDOS)` | Vira 0, 1, 2, 0, 1, 2...: corredor, cantina, parquinho, corredor... |
| `numero % 12` | Posição dentro do fundo, **contando de baixo** (0 a 11) |
| `12 - 1 - ...` | Vira a contagem **de cima** (a lista `FAIXAS_FUNDO` é de cima para baixo) |

Ex: faixa 13 → `13 // 12 = 1` → cantina; `13 % 12 = 1` → `linha = 10` → `FAIXAS_FUNDO[1][10]` = `"livre"`.

💡 `len(FUNDOS)` em vez de `3`: se entrar um 4º mapa, o código continua certo.

```python
def criar_obstaculos_da_faixa(obstaculos, y, numero):
    if tipo_da_faixa(numero) == "livre":
        return
    nome = random.choice(OBSTACULOS_DO_MAPA[qual_fundo(numero)])
    tipo = TIPOS_OBSTACULO[nome]
    direcao = random.choice([1, -1])
    velocidade = random.choice(tipo["velocidades"]) * direcao
    x = random.randint(0, 250)
    for i in range(tipo["quantos"]):
        imagem = random.choice(tipo["imagens"])
        obstaculos.append(criar_obstaculo(x + i * 300, y, velocidade, tipo["largura"], imagem))
```

**Sorteia quem anda numa faixa de corredor.**
| Linha | O que faz |
|---|---|
| `if ... == "livre": return` | Faixa livre não tem obstáculo: sai da função |
| `OBSTACULOS_DO_MAPA[qual_fundo(numero)]` | A lista de quem pode aparecer **no mapa dessa faixa** |
| `nome = random.choice(...)` | Sorteia um nome, ex: `"merendeira"` |
| `tipo = TIPOS_OBSTACULO[nome]` | Pega as características dele (Bloco 4) |
| `random.choice([1, -1])` | Sorteia o sentido: 1 = direita, -1 = esquerda |
| `random.choice(tipo["velocidades"]) * direcao` | Velocidade dele, no sentido sorteado |
| `for i in range(tipo["quantos"]):` | Cria 1 ou 2 |
| `x + i * 300` | O 1º em `x`, o 2º 300 pixels depois |
| `obstaculos.append(...)` | Coloca o obstáculo novo no fim da lista |

```python
def montar_mapa_inicial(obstaculos, mundo):
    mundo["linhas_roladas"] = 0
    obstaculos.clear()
    for numero in range(FAIXAS_POR_FUNDO):
        y = ALTURA - CASA - numero * CASA
        criar_obstaculos_da_faixa(obstaculos, y, numero)
```

**Começo de cada partida: monta as 12 faixas da tela.**
| Linha | O que faz |
|---|---|
| `linhas_roladas = 0` | O mundo volta para a largada |
| `obstaculos.clear()` | Esvazia a lista (apaga os obstáculos da partida anterior) |
| `for numero in range(12):` | Faixa 0 (embaixo) até 11 (em cima) |
| `y = ALTURA - CASA - numero * CASA` | Faixa 0 → y = 550, faixa 1 → 500, ..., faixa 11 → 0 |

```python
def rolar_mundo(obstaculos, mundo):
    mundo["linhas_roladas"] = mundo["linhas_roladas"] + 1
    for obstaculo in obstaculos:
        obstaculo["y"] = obstaculo["y"] + CASA
    for obstaculo in obstaculos[:]:
        if obstaculo["y"] >= ALTURA:
            obstaculos.remove(obstaculo)
    numero_da_faixa_nova = mundo["linhas_roladas"] + FAIXAS_POR_FUNDO - 1
    criar_obstaculos_da_faixa(obstaculos, 0, numero_da_faixa_nova)
```

**O coração do mapa infinito: o mundo desce uma faixa.**
| Linha | O que faz |
|---|---|
| `linhas_roladas + 1` | Conta mais uma faixa descida (o fundo usa isso para se desenhar mais embaixo) |
| `obstaculo["y"] + CASA` | Todos os obstáculos descem 50 pixels |
| `for obstaculo in obstaculos[:]:` | Percorre uma **cópia** da lista (`[:]`): remover itens da lista que está sendo percorrida faz o `for` pular itens |
| `if y >= ALTURA: remove` | Quem saiu por baixo da tela é apagado |
| `numero_da_faixa_nova` | O número da faixa que acabou de aparecer **em cima** (y = 0) |
| `criar_obstaculos_da_faixa(..., 0, ...)` | Sorteia os obstáculos dela |

💡 A lista nunca cresce sem parar: entra uma faixa em cima, sai uma embaixo.

```python
def desenhar_fundo(tela, mundo):
    primeiro = mundo["linhas_roladas"] // FAIXAS_POR_FUNDO
    for numero in [primeiro, primeiro + 1]:
        y = mundo["linhas_roladas"] * CASA - numero * ALTURA
        tela.blit(FUNDOS[numero % len(FUNDOS)], (0, y))
```

**Desenha os 2 pedaços de fundo que aparecem na tela.**
| Linha | O que faz |
|---|---|
| `primeiro = linhas_roladas // 12` | Qual cópia de fundo está embaixo na tela |
| `for numero in [primeiro, primeiro + 1]` | A tela mostra no máximo 2 fundos: o de baixo e o de cima dele |
| `y = linhas_roladas * CASA - numero * ALTURA` | Cada faixa rolada empurra o fundo 50 px para baixo; cada cópia fica 600 px acima da anterior |
| `FUNDOS[numero % len(FUNDOS)]` | Escolhe corredor, cantina ou parquinho |

Ex: 15 faixas roladas → `primeiro = 1`: a cantina fica em `y = 750 - 600 = 150` e o parquinho em `y = 750 - 1200 = -450` (a parte acima da tela não aparece).

# 🧱 BLOCO 6 – Colisão
**Objetivo:** saber se o personagem está encostando em algum obstáculo.

```python
def bateu(personagem, obstaculos):
    retangulo_personagem = pygame.Rect(personagem["x"], personagem["y"], CASA, CASA)
    for obstaculo in obstaculos:
        retangulo_obstaculo = pygame.Rect(obstaculo["x"], obstaculo["y"], obstaculo["largura"], CASA)
        if retangulo_personagem.colliderect(retangulo_obstaculo):
            return True
    return False
```

**Responde True (bateu) ou False (não bateu).**
| Linha | O que faz |
|---|---|
| `pygame.Rect(x, y, largura, altura)` | Cria uma "caixa" com a posição e o tamanho do personagem |
| `for obstaculo in obstaculos:` | Testa um obstáculo por vez |
| `pygame.Rect(..., obstaculo["largura"], CASA)` | Caixa do obstáculo, com a largura dele (50 ou 100) |
| `colliderect(...)` | As duas caixas estão uma em cima da outra? |
| `return True` | Achou uma batida: devolve True e **sai da função na hora** |
| `return False` | Fica **fora** do `for`: só chega aqui se testou todos e nenhum bateu |

💡 **Cuidado com a indentação:** se o `return False` estivesse dentro do `for`, a função pararia no primeiro obstáculo. Só encostar a borda **não** conta como batida.

3. Para que serve o `pygame.Rect`? <details><summary>R</summary>Representar uma caixa (posição e tamanho) e usar `colliderect` para ver se duas caixas se sobrepõem.</details>

# 🧱 BLOCO 7 – Pontuação e recorde
**Objetivo:** contar os pulos (faixas novas alcançadas), guardar o recorde e mostrar na tela.

```python
placar = {"pontos": 0, "recorde": 0}

def atualizar_pontos(placar, personagem):
    if personagem["faixa"] > placar["pontos"]:
        placar["pontos"] = personagem["faixa"]
        return True
    return False

def zerar_pontos(placar):
    placar["pontos"] = 0

def guardar_recorde(placar):
    if placar["pontos"] > placar["recorde"]:
        placar["recorde"] = placar["pontos"]
```

| Linha | O que faz |
|---|---|
| `placar = {...}` | Os pulos da partida e o recorde (maior número de pulos desde que o jogo abriu) |
| `if faixa > pontos` | Chegou numa faixa mais alta que todas da partida? |
| `pontos = faixa` | Os pontos são a faixa mais alta alcançada |
| `return True` | Avisa o loop que é faixa **nova** (aí a barra de aperto esvazia) |
| `zerar_pontos` | Zera os pontos para outra partida. O recorde **fica** |
| `guardar_recorde` | No game over: se a partida foi melhor, vira o recorde |

💡 Voltar e subir de novo numa faixa já visitada **não** dá ponto: a faixa não é maior que os pontos.

```python
def desenhar_sombra(tela, y, altura, quanto):
    sombra = pygame.Surface((LARGURA, altura))
    sombra.set_alpha(quanto)
    tela.blit(sombra, (0, y))

def desenhar_pontos(tela, placar):
    desenhar_sombra(tela, 0, 40, 150)
    texto = fonte_pequena.render("Pulos: " + str(placar["pontos"]) + "    Recorde: " + str(placar["recorde"]), True, BRANCO)
    tela.blit(texto, (10, 12))
```

| Linha | O que faz |
|---|---|
| `pygame.Surface((LARGURA, altura))` | Cria uma "folha" retangular (começa preta) |
| `set_alpha(quanto)` | Deixa transparente: 0 = invisível, 255 = preto total |
| `desenhar_sombra(tela, 0, 40, 150)` | Tarja escura no alto da tela, para o texto aparecer bem |
| `str(placar["pontos"])` | Transforma o número em texto (não dá para juntar texto + número com `+`) |
| `fonte.render(texto, True, BRANCO)` | Transforma o texto numa **imagem**. `True` = bordas suaves |
| `tela.blit(texto, (10, 12))` | **Cola** a imagem do texto na tela |

💡 O pygame não escreve texto direto: primeiro vira imagem (`render`), depois é colado (`blit`).

# 🧱 BLOCO 8 – Barra de aperto
**Objetivo:** uma barra que enche com o tempo (7 segundos). Cheia = game over. Cada faixa nova esvazia ela.

```python
APERTO_MAXIMO = 100
VELOCIDADE_APERTO = APERTO_MAXIMO / (7 * 60)

barra = {"aperto": 0}

def encher_barra(barra):
    barra["aperto"] = barra["aperto"] + VELOCIDADE_APERTO

def barra_cheia(barra):
    return barra["aperto"] >= APERTO_MAXIMO

def esvaziar_barra(barra):
    barra["aperto"] = 0
```

| Linha | O que faz |
|---|---|
| `APERTO_MAXIMO = 100` | Quando o aperto chega em 100, a barra está cheia |
| `100 / (7 * 60)` | 7 segundos × 60 quadros = 420 quadros. Enche ≈ 0,24 por quadro |
| `barra = {"aperto": 0}` | Mais um dicionário, igual ao personagem e ao placar |
| `encher_barra` | Chamada **todo quadro**: soma um pouquinho |
| `barra_cheia` | Devolve True quando chegou no máximo (`>=` porque a soma pode passar um pouco de 100) |
| `esvaziar_barra` | Volta para zero (faixa nova ou partida nova) |

```python
def desenhar_barra(tela, barra):
    largura_cheia = int(barra["aperto"] * 2)
    pygame.draw.rect(tela, VERMELHO, (390, 10, largura_cheia, 20))
    pygame.draw.rect(tela, BRANCO, (390, 10, 200, 20), 2)
```

| Linha | O que faz |
|---|---|
| `int(aperto * 2)` | 100 de aperto = 200 pixels. `int` porque o desenho precisa de número inteiro |
| `draw.rect(tela, VERMELHO, (x, y, largura, altura))` | A parte cheia, que cresce |
| `draw.rect(..., 2)` | O último número é a **espessura**: desenha só o contorno (a moldura) |

2. Por que `barra_cheia` usa `>=` e não `==`? <details><summary>R</summary>A barra soma números quebrados e pode passar um pouco de 100 sem nunca ser exatamente 100.</details>

# 🧱 BLOCO 11 – Loop principal
**Objetivo:** juntar tudo. O jogo é um `while` que repete 60 vezes por segundo. Cada volta = 1 quadro.

🔁 **Ordem de todo quadro:** 1) ler teclas → 2) atualizar → 3) desenhar → 4) mostrar → 5) esperar

```python
rodando = True
while rodando:

    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
        if evento.type == pygame.KEYDOWN and evento.key == pygame.K_ESCAPE:
            rodando = False
        if evento.type == pygame.KEYDOWN:
            if estado == "jogo":
                if mover_personagem(personagem, evento.key):
                    rolar_mundo(obstaculos, mundo)
            elif evento.key == pygame.K_SPACE:
                reiniciar_jogo(personagem, placar, barra, obstaculos, mundo)
                estado = "jogo"
```

**1) Ler o que o jogador fez.**
| Linha | O que faz |
|---|---|
| `while rodando:` | Repete até `rodando` virar False |
| `pygame.event.get()` | Lista do que aconteceu desde o último quadro (teclas, clique no X...) |
| `pygame.QUIT` | Clicou no X: o `while` para |
| `evento.key == pygame.K_ESCAPE` | Apertou ESC: o `while` também para. Em tela cheia não aparece o X, então o ESC é o jeito de sair |
| `pygame.KEYDOWN` | Uma tecla foi **apertada** (acontece uma vez por apertada) |
| `if mover_personagem(...): rolar_mundo(...)` | No jogo: move e, se ele estava no limite, o mundo desce |
| `elif evento.key == pygame.K_SPACE:` | No início ou no fim: ESPAÇO começa uma partida nova |

```python
    if estado == "jogo":
        mover_obstaculos(obstaculos)

        if personagem["animacao"] != "batida" and bateu(personagem, obstaculos):
            som_batida.play()
            comecar_animacao(personagem, "batida", TEMPO_BATIDA)

        atualizar_animacao(personagem)

        if terminou_batida(personagem):
            guardar_recorde(placar)
            estado = "fim"
            frase_fim = random.choice(FRASES_BATIDA)

        if atualizar_pontos(placar, personagem):
            esvaziar_barra(barra)

        encher_barra(barra)
        if barra_cheia(barra):
            guardar_recorde(placar)
            estado = "fim"
            frase_fim = random.choice(FRASES_FIM)
```

**2) Atualizar** (só quando `estado == "jogo"`; no fim, tudo fica parado).
| Linha | O que faz |
|---|---|
| `mover_obstaculos` | Todo quadro, mesmo sem tecla apertada |
| `animacao != "batida" and bateu(...)` | Bateu agora? (se já está batendo, não conta de novo, senão o som tocaria 60 vezes) |
| `comecar_animacao(..., "batida", ...)` | 1 segundo tonto, com estrelinhas, sem poder andar |
| `if terminou_batida(...)` | A batida acabou: guarda o recorde e vai para o **fim** |
| `if atualizar_pontos(...): esvaziar_barra(...)` | Faixa nova: ganha ponto **e** a barra zera |
| `encher_barra` / `barra_cheia` | A barra enche; se encher: recorde e **fim** |

```python
    if estado == "inicio":
        desenhar_tela_inicio(tela)
    else:
        desenhar_fundo(tela, mundo)
        desenhar_obstaculos(tela, obstaculos)
        desenhar_personagem(tela, personagem)
        desenhar_pontos(tela, placar)
        desenhar_barra(tela, barra)
        if estado == "fim":
            desenhar_tela_fim(tela, placar, frase_fim)

    pygame.display.update()
    relogio.tick(60)

pygame.quit()
```

**3) Desenhar, 4) mostrar, 5) esperar.**
| Linha | O que faz |
|---|---|
| `else:` | Vale para `"jogo"` **e** `"fim"`: desenha a escola |
| Ordem dos `desenhar_...` | De trás para frente: fundo → obstáculos → personagem → placar e barra por cima |
| `if estado == "fim":` | No fim, a tela de GAME OVER vem por cima do jogo parado |
| `pygame.display.update()` | **Mostra** o que foi desenhado. Sem ele, a janela fica preta |
| `relogio.tick(60)` | Espera o suficiente para dar 60 quadros por segundo (mesma velocidade em qualquer PC) |
| `pygame.quit()` | Fora do `while`: desliga o pygame quando o jogo fecha |

💡 **A ordem do desenho importa:** o que é desenhado depois fica **por cima**. Se o fundo viesse por último, cobriria tudo.

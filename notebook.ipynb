# 🧱 BLOCO 1 – Janela e loop principal
**Objetivo:** abrir uma janela verde que fica aberta até clicar no X.

```python
import pygame

pygame.init()
```

**Liga o pygame.**
| Linha | O que faz |
|---|---|
| `import pygame` | Traz a biblioteca para o programa |
| `pygame.init()` | Liga as partes do pygame (vídeo, teclado, som). Sempre logo após o import |

```python
LARGURA = 600
ALTURA = 600

tela = pygame.display.set_mode((LARGURA, ALTURA))
pygame.display.set_caption("Segura Aí!")
```

**Cria a janela.** A `tela` é onde tudo vai ser desenhado.
| Linha | O que faz |
|---|---|
| `LARGURA = 600` / `ALTURA = 600` | Tamanho da janela em pixels. MAIÚSCULO = constante (valor que não muda) |
| `set_mode((LARGURA, ALTURA))` | Cria a janela e guarda em `tela`. Parênteses duplos: o tamanho vai como um par `(600, 600)` |
| `set_caption(...)` | Título da janela |

📍 **Coordenadas:** `(0, 0)` é o **canto superior esquerdo**. x cresce para a direita, **y cresce para baixo**.

```python
relogio = pygame.time.Clock()

VERDE = (100, 200, 100)
```

| Linha | O que faz |
|---|---|
| `relogio = pygame.time.Clock()` | Relógio para controlar a velocidade do jogo (usado no `tick`) |
| `VERDE = (100, 200, 100)` | Cor em RGB: (Vermelho, Verde, Azul), cada um de 0 a 255. Ex: `(0,0,0)` preto, `(255,255,255)` branco |

```python
rodando = True
while rodando:

    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
```

**Loop principal:** o jogo é um `while` que repete ~60 vezes por segundo. Cada volta = 1 quadro (frame).
| Linha | O que faz |
|---|---|
| `rodando = True` | Controla se o jogo continua |
| `while rodando:` | Repete enquanto `rodando` for `True` |
| `for evento in pygame.event.get():` | Percorre a lista do que o jogador fez (teclas, cliques, X) |
| `if evento.type == pygame.QUIT:` | O jogador clicou no X? |
| `rodando = False` | Então o `while` para |

```python
    tela.fill(VERDE)

    pygame.display.update()

    relogio.tick(60)
```

*(continua dentro do `while`)*
| Linha | O que faz |
|---|---|
| `tela.fill(VERDE)` | Pinta a tela toda. Feito todo quadro para **apagar** o anterior (senão fica rastro) |
| `pygame.display.update()` | **Mostra** o que foi desenhado. Sem ele, a janela fica preta |
| `relogio.tick(60)` | Limita a 60 FPS, para ter a mesma velocidade em qualquer PC |

🔁 **Ordem de todo quadro:** eventos → atualizar → desenhar → `update` → `tick`

```python
pygame.quit()
```

*(fora do `while`)* Desliga o pygame e fecha a janela quando o loop acaba.

# 🧱 BLOCO 2 – Personagem andando com as setas
**Objetivo:** um quadrado azul (o personagem) que anda **uma casa por vez** com as setas.

🗂️ **Como organizamos o jogo (sem classes):** os **dados** ficam em **dicionários** e as **ações** em **funções**.

```python
AZUL = (50, 100, 255)

CASA = 50

personagem = {"x": 250, "y": 550}
```

| Linha | O que faz |
|---|---|
| `AZUL = (50, 100, 255)` | Cor do personagem |
| `CASA = 50` | Tamanho de cada "casa" do chão. O personagem anda 50 pixels por vez (como num tabuleiro). A tela 600×600 vira 12×12 casas |
| `personagem = {"x": 250, "y": 550}` | O personagem é um **dicionário** com sua posição. `x=250` ≈ meio da tela, `y=550` = última linha (lá embaixo) |

💡 `x` e `y` são o **canto superior esquerdo** do quadrado.

```python
def mover_personagem(personagem, tecla):
    if tecla == pygame.K_UP and personagem["y"] > 0:
        personagem["y"] = personagem["y"] - CASA
    if tecla == pygame.K_DOWN and personagem["y"] < ALTURA - CASA:
        personagem["y"] = personagem["y"] + CASA
    if tecla == pygame.K_LEFT and personagem["x"] > 0:
        personagem["x"] = personagem["x"] - CASA
    if tecla == pygame.K_RIGHT and personagem["x"] < LARGURA - CASA:
        personagem["x"] = personagem["x"] + CASA
```

**Recebe a tecla apertada e anda uma casa naquela direção.**
| Linha | O que faz |
|---|---|
| `def mover_personagem(personagem, tecla):` | Cria a função. Recebe o dicionário do personagem e a tecla |
| `if tecla == pygame.K_UP and personagem["y"] > 0:` | Apertou ↑ **e** não está no topo? |
| `personagem["y"] = personagem["y"] - CASA` | Sobe: **diminui** o y (lembre: y cresce para baixo) |
| ↓ | Aumenta o y, se não estiver na última linha (`ALTURA - CASA` = 550) |
| ← / → | Diminui / aumenta o x, com o mesmo cuidado nas bordas |

💡 **Por que `ALTURA - CASA` e não `ALTURA`?** Como x/y é o canto de cima do quadrado, se y fosse 600 ele estaria **fora** da tela. A última posição possível é 600 − 50 = 550.

💡 **A função muda o dicionário original.** Quando passamos um dicionário para uma função e alteramos ele lá dentro, a mudança vale fora também. Por isso não precisa de `return`.

```python
def desenhar_personagem(tela, personagem):
    pygame.draw.rect(tela, AZUL, (personagem["x"], personagem["y"], CASA, CASA))
```

| Linha | O que faz |
|---|---|
| `pygame.draw.rect(tela, cor, (x, y, largura, altura))` | Desenha um retângulo na `tela`. Aqui: na posição do personagem, com 50×50 |

```python
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
        if evento.type == pygame.KEYDOWN:
            mover_personagem(personagem, evento.key)

    tela.fill(VERDE)
    desenhar_personagem(tela, personagem)
```

**Mudanças dentro do loop principal:**
| Linha | O que faz |
|---|---|
| `if evento.type == pygame.KEYDOWN:` | Alguma tecla foi **apertada**? |
| `mover_personagem(personagem, evento.key)` | Chama a função passando **qual** tecla foi (`evento.key`) |
| `desenhar_personagem(tela, personagem)` | Desenha o personagem. Vem **depois** do `fill`, senão o fundo cobriria ele |

💡 **KEYDOWN** acontece **uma vez** por apertada. Por isso o personagem anda uma casa, e não sai deslizando enquanto a tecla está segurada.

# 🧱 BLOCO 3 – Obstáculos andando no corredor
**Objetivo:** pessoas e carrinhos que andam sozinhos pelo corredor, de um lado para o outro.

💡 Todo obstáculo é **o mesmo tipo de dicionário**; só mudam os valores (velocidade, largura, cor).

```python
LARANJA = (255, 150, 50)
VERMELHO = (220, 50, 50)

def criar_obstaculo(x, y, velocidade, largura, cor):
    return {"x": x, "y": y, "velocidade": velocidade, "largura": largura, "cor": cor}
```

**Fábrica de obstáculos:** recebe os valores e **devolve** (`return`) um dicionário pronto.
| Linha | O que faz |
|---|---|
| `LARANJA` / `VERMELHO` | Cores novas: aluno (laranja) e carrinho de limpeza (vermelho) |
| `def criar_obstaculo(...)` | Função com 5 parâmetros |
| `return {...}` | Monta o dicionário com os valores recebidos e devolve |

💡 Assim não precisamos escrever o dicionário inteiro toda vez.

```python
obstaculos = [
    criar_obstaculo(0, 450, 2, 50, LARANJA),
    criar_obstaculo(300, 450, 2, 50, LARANJA),
    criar_obstaculo(100, 350, -5, 100, VERMELHO),
    criar_obstaculo(400, 250, 3, 50, LARANJA),
    criar_obstaculo(200, 150, -6, 100, VERMELHO),
]
```

**Lista com todos os obstáculos** (uma lista de dicionários).
| Valor | Significado |
|---|---|
| `y` (450, 350...) | Em qual faixa ele anda. Sempre múltiplo de 50 para encaixar nas casas |
| `velocidade` | Pixels por quadro. **Positivo → direita**, **negativo → esquerda** |
| `largura` | 50 = aluno (1 casa), 100 = carrinho (2 casas) |

Ex: `criar_obstaculo(100, 350, -5, 100, VERMELHO)` = carrinho na faixa y=350, rápido, indo para a esquerda.

```python
def mover_obstaculos(obstaculos):
    for obstaculo in obstaculos:
        obstaculo["x"] = obstaculo["x"] + obstaculo["velocidade"]
        if obstaculo["x"] > LARGURA:
            obstaculo["x"] = -obstaculo["largura"]
        if obstaculo["x"] < -obstaculo["largura"]:
            obstaculo["x"] = LARGURA
```

**Anda com todos e faz quem sai da tela voltar pelo outro lado.**
| Linha | O que faz |
|---|---|
| `for obstaculo in obstaculos:` | Passa por cada dicionário da lista |
| `x = x + velocidade` | Anda. Se a velocidade é negativa, o x diminui (vai para a esquerda) |
| `if x > LARGURA:` → `x = -largura` | Saiu pela **direita**: reaparece escondido logo antes da borda **esquerda** |
| `if x < -largura:` → `x = LARGURA` | Saiu pela **esquerda** (sumiu por inteiro): reaparece na borda **direita** |

💡 Usamos `-largura` para ele entrar na tela **aos poucos**, e não "piscar" já dentro dela.

```python
def desenhar_obstaculos(tela, obstaculos):
    for obstaculo in obstaculos:
        pygame.draw.rect(tela, obstaculo["cor"], (obstaculo["x"], obstaculo["y"], obstaculo["largura"], CASA))
```

Igual ao `desenhar_personagem`, mas para cada obstáculo, usando a **cor** e a **largura** dele. A altura é sempre `CASA` (50).

```python
    mover_obstaculos(obstaculos)

    tela.fill(VERDE)
    desenhar_obstaculos(tela, obstaculos)
    desenhar_personagem(tela, personagem)
```

**Mudanças dentro do loop principal:**
| Linha | O que faz |
|---|---|
| `mover_obstaculos(obstaculos)` | Fica **fora** do `for evento`: os obstáculos andam em **todo quadro**, mesmo sem apertar tecla |
| `desenhar_obstaculos(...)` | Depois do `fill` e **antes** do personagem, para o personagem aparecer por cima |

🔁 Agora o loop tem as 3 etapas: **eventos → atualizar → desenhar**.

2. Por que `mover_obstaculos` fica fora do `for evento`? <details><summary>R</summary>Para andarem todo quadro, não só quando há tecla apertada.</details>
3. Por que `x = -largura` e não `x = 0`? <details><summary>R</summary>Para entrar na tela aos poucos, começando escondido.</details>
4. O que `criar_obstaculo` devolve? <details><summary>R</summary>Um dicionário com x, y, velocidade, largura e cor.</details>

# 🧱 BLOCO 4 – Colisão
**Objetivo:** se o personagem encostar num obstáculo, ele volta para o começo.

```python
def bateu(personagem, obstaculos):
    retangulo_personagem = pygame.Rect(personagem["x"], personagem["y"], CASA, CASA)
    for obstaculo in obstaculos:
        retangulo_obstaculo = pygame.Rect(obstaculo["x"], obstaculo["y"], obstaculo["largura"], CASA)
        if retangulo_personagem.colliderect(retangulo_obstaculo):
            return True
    return False
```

**Responde True (bateu) ou False (não bateu).**
| Linha | O que faz |
|---|---|
| `pygame.Rect(x, y, largura, altura)` | Cria uma "caixa" com a posição e o tamanho do personagem |
| `for obstaculo in obstaculos:` | Testa um obstáculo por vez |
| `retangulo_obstaculo = pygame.Rect(...)` | Caixa do obstáculo (com a largura dele: 50 ou 100) |
| `if ...colliderect(...)` | As duas caixas estão uma em cima da outra? |
| `return True` | Achou uma batida: devolve True e **sai da função na hora** |
| `return False` | Fica **fora** do `for`: só chega aqui se testou todos e nenhum bateu |

💡 **Cuidado com a indentação:** se o `return False` estivesse dentro do `for`, a função pararia logo no primeiro obstáculo.

💡 Só encostar a borda (um do lado do outro) **não** conta como batida, só quando um entra no outro.

```python
def voltar_ao_comeco(personagem):
    personagem["x"] = 250
    personagem["y"] = 550
```

Coloca o personagem de volta na posição inicial (a mesma do Bloco 2). Igual ao `mover_personagem`: muda o dicionário original, sem `return`.

```python
    mover_obstaculos(obstaculos)

    if bateu(personagem, obstaculos):
        voltar_ao_comeco(personagem)
```

**Mudança no loop principal**, na etapa "atualizar":
| Linha | O que faz |
|---|---|
| `if bateu(personagem, obstaculos):` | Usa o True/False que a função devolve direto no `if` |
| `voltar_ao_comeco(personagem)` | Se bateu, volta para o começo |

💡 Fica depois de `mover_obstaculos` e fora do `for evento`: é testado **todo quadro**, porque o obstáculo pode bater no personagem parado.

# 🧱 BLOCO 5 – Pontuação
**Objetivo:** ganhar 1 ponto por faixa nova alcançada e mostrar os pontos na tela. Bateu? Os pontos voltam a 0.

```python
PRETO = (0, 0, 0)

fonte = pygame.font.Font(None, 40)

placar = {"pontos": 0, "menor_y": 550}
```

| Linha | O que faz |
|---|---|
| `PRETO = (0, 0, 0)` | Cor do texto (fica junto das outras cores, no Bloco 1) |
| `pygame.font.Font(None, 40)` | Cria uma fonte para escrever na tela. `None` = fonte padrão do pygame, `40` = tamanho |
| `placar = {...}` | O placar é um **dicionário**, igual ao personagem |
| `"pontos": 0` | Quantos pontos o jogador tem |
| `"menor_y": 550` | A faixa **mais alta** que o personagem já alcançou. Começa em 550 (a linha de partida) |

💡 Como **y diminui quando sobe**, a faixa mais alta é a de **menor** y.

```python
def atualizar_pontos(placar, personagem):
    if personagem["y"] < placar["menor_y"]:
        placar["menor_y"] = personagem["y"]
        placar["pontos"] = placar["pontos"] + 1
```

**Dá 1 ponto quando o personagem chega numa faixa onde nunca esteve.**
| Linha | O que faz |
|---|---|
| `if personagem["y"] < placar["menor_y"]:` | O personagem está mais alto do que o recorde? |
| `placar["menor_y"] = personagem["y"]` | Guarda o novo recorde |
| `placar["pontos"] = ... + 1` | Ganha 1 ponto |

💡 Sem o `menor_y`, dava para ficar subindo e descendo na mesma faixa e ganhar pontos infinitos.

```python
def zerar_pontos(placar):
    placar["pontos"] = 0
    placar["menor_y"] = 550
```

Volta o placar ao começo: 0 pontos e recorde na linha de partida. Muda o dicionário original, sem `return` (igual ao `voltar_ao_comeco`).

```python
def desenhar_pontos(tela, placar):
    texto = fonte.render("Pontos: " + str(placar["pontos"]), True, PRETO)
    tela.blit(texto, (10, 10))
```

**Escreve os pontos no canto de cima da tela.**
| Linha | O que faz |
|---|---|
| `str(placar["pontos"])` | Transforma o número em texto. Não dá para juntar texto + número com `+` |
| `fonte.render(texto, True, PRETO)` | Transforma o texto numa **imagem**. `True` = bordas das letras suaves |
| `tela.blit(texto, (10, 10))` | **Cola** a imagem na tela, na posição x=10, y=10 (canto superior esquerdo) |

💡 O pygame não escreve texto direto na tela: primeiro vira imagem (`render`), depois é colado (`blit`).

```python
    if bateu(personagem, obstaculos):
        voltar_ao_comeco(personagem)
        zerar_pontos(placar)

    atualizar_pontos(placar, personagem)

    tela.fill(VERDE)
    desenhar_obstaculos(tela, obstaculos)
    desenhar_personagem(tela, personagem)
    desenhar_pontos(tela, placar)
```

**Mudanças no loop principal:**
| Linha | O que faz |
|---|---|
| `zerar_pontos(placar)` | Dentro do `if bateu`: além de voltar ao começo, perde os pontos |
| `atualizar_pontos(placar, personagem)` | Na etapa "atualizar", todo quadro: confere se chegou numa faixa nova |
| `desenhar_pontos(tela, placar)` | **Por último** no desenho, para o texto ficar por cima de tudo |

## 📝 Perguntas de prova – Bloco 5
1. Para que serve o `menor_y`? <details><summary>R</summary>Guardar a faixa mais alta já alcançada, para só ganhar ponto em faixa nova.</details>
2. Por que é `<` e não `>` no `if` de `atualizar_pontos`? <details><summary>R</summary>Porque y diminui quando o personagem sobe.</details>
3. Por que usamos `str()` em `desenhar_pontos`? <details><summary>R</summary>Para transformar o número em texto e poder juntar com "Pontos: ".</details>
4. Qual a diferença entre `render` e `blit`? <details><summary>R</summary>`render` transforma o texto em imagem; `blit` cola essa imagem na tela.</details>
5. Por que `desenhar_pontos` é chamado por último? <details><summary>R</summary>Para o texto ficar por cima do fundo, dos obstáculos e do personagem.</details>

# 🧱 BLOCO 6 – Barra de aperto
**Objetivo:** uma barra que enche com o tempo. Se encher, o jogador recomeça do zero (por enquanto; a tela de fim vem depois).

```python
APERTO_MAXIMO = 100
VELOCIDADE_APERTO = 0.1

barra = {"aperto": 0}
```

| Linha | O que faz |
|---|---|
| `APERTO_MAXIMO = 100` | Valor em que a barra fica cheia |
| `VELOCIDADE_APERTO = 0.1` | Quanto a barra enche **por quadro**. 100 / 0.1 = 1000 quadros ≈ 17 segundos (a 60 FPS) |
| `barra = {"aperto": 0}` | A barra é um **dicionário**, igual ao personagem e ao placar. Começa vazia |

💡 Para deixar o jogo mais difícil, é só aumentar `VELOCIDADE_APERTO`.

```python
def encher_barra(barra):
    barra["aperto"] = barra["aperto"] + VELOCIDADE_APERTO


def barra_cheia(barra):
    return barra["aperto"] >= APERTO_MAXIMO


def esvaziar_barra(barra):
    barra["aperto"] = 0
```

| Função | O que faz |
|---|---|
| `encher_barra` | Soma um pouquinho no aperto. É chamada **todo quadro** |
| `barra_cheia` | Devolve **True** ou **False** (igual ao `bateu` do Bloco 4), por isso pode ir direto num `if` |
| `esvaziar_barra` | Volta o aperto para 0 |

💡 Usamos `>=` e não `==`: somando 0.1 várias vezes, o computador pode chegar em 100.00001 e **nunca** ser exatamente 100.

```python
def desenhar_barra(tela, barra):
    largura_cheia = int(barra["aperto"] * 2)
    pygame.draw.rect(tela, VERMELHO, (390, 10, largura_cheia, 20))
    pygame.draw.rect(tela, PRETO, (390, 10, 200, 20), 2)
```

**Desenha a barra no canto superior direito.**
| Linha | O que faz |
|---|---|
| `largura_cheia = int(aperto * 2)` | Converte o aperto em pixels: 100 de aperto = 200 pixels. `int` porque pixel é número inteiro |
| 1º `draw.rect` (VERMELHO) | O **preenchimento**: cresce conforme o aperto |
| 2º `draw.rect` (PRETO, ..., `2`) | O **contorno**. O último número (`2`) é a espessura da borda: desenha só a borda, sem pintar o meio |

💡 A barra vai de x=390 até x=590 e fica na faixa de cima, onde não passa obstáculo. O contorno vem **depois** para ficar por cima do vermelho.

```python
    atualizar_pontos(placar, personagem)

    encher_barra(barra)
    if barra_cheia(barra):
        voltar_ao_comeco(personagem)
        zerar_pontos(placar)
        esvaziar_barra(barra)

    tela.fill(VERDE)
    desenhar_obstaculos(tela, obstaculos)
    desenhar_personagem(tela, personagem)
    desenhar_pontos(tela, placar)
    desenhar_barra(tela, barra)
```

**Mudanças no loop principal:**
| Linha | O que faz |
|---|---|
| `encher_barra(barra)` | Na etapa "atualizar", **todo quadro**: a barra enche com o tempo, mesmo parado |
| `if barra_cheia(barra):` | Se encheu: volta ao começo, zera os pontos e esvazia a barra |

# 🧱 BLOCO 8 – Telas (início, jogo e fim)
**Objetivo:** o jogo passa a ter 3 telas. ESPAÇO começa e reinicia; a barra cheia dá **game over** com uma frase sorteada; chegar ao banheiro é **vitória**.

🗂️ **Ideia principal:** uma variável `estado` guarda **em qual tela estamos**. O loop olha para ela para decidir o que ler, o que atualizar e o que desenhar.

```python
import pygame
import random
```

| Linha | O que faz |
|---|---|
| `import random` | Traz a biblioteca de sorteios. Fica lá no topo, junto do `import pygame` |

```python
estado = "inicio"

FRASES_FIM = [
    "Não deu tempo...",
    "Ih, molhou tudo!",
    "Faltou só um pouquinho!",
    "Alguém chama a limpeza!",
]

frase_fim = ""
```

**As variáveis das telas.**
| Linha | O que faz |
|---|---|
| `estado = "inicio"` | Em qual tela estamos: `"inicio"`, `"jogo"` ou `"fim"`. O jogo **abre** na tela de início |
| `FRASES_FIM = [...]` | Lista de frases de game over. MAIÚSCULO = constante (a lista não muda) |
| `frase_fim = ""` | A frase que vai aparecer na tela de fim. Começa vazia e recebe um valor quando o jogo acaba |

```python
def reiniciar_jogo(personagem, placar, barra):
    voltar_ao_comeco(personagem)
    zerar_pontos(placar)
    esvaziar_barra(barra)


def chegou_no_banheiro(personagem):
    return personagem["y"] == 0
```

**Duas funções curtas.**
| Linha | O que faz |
|---|---|
| `def reiniciar_jogo(...)` | Junta 3 funções dos blocos anteriores para começar uma partida do zero |
| `voltar_ao_comeco(personagem)` | Personagem volta para baixo (Bloco 4) |
| `zerar_pontos(placar)` | Pontos voltam a 0 (Bloco 5) |
| `esvaziar_barra(barra)` | Barra volta a 0 (Bloco 6) |
| `return personagem["y"] == 0` | Devolve `True` se o personagem está na faixa de cima (o banheiro), senão `False` |

💡 `personagem["y"] == 0` já **é** `True` ou `False`, então dá para devolver direto, sem `if`.

```python
def desenhar_tela_inicio(tela):
    tela.fill(AZUL_CLARO)
    titulo = fonte.render("Segura Aí!", True, PRETO)
    instrucao = fonte_pequena.render("Aperte ESPAÇO para começar", True, PRETO)
    tela.blit(titulo, (220, 250))
    tela.blit(instrucao, (180, 320))
```

**Tela de início.**

```python
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
        if evento.type == pygame.KEYDOWN:
            if estado == "jogo":
                mover_personagem(personagem, evento.key)
            elif evento.key == pygame.K_SPACE:
                reiniciar_jogo(personagem, placar, barra)
                estado = "jogo"
```

# 🧱 BLOCO 1 – A janela e o loop principal

**Objetivo do bloco:** abrir uma janela verde que fica aberta até o jogador clicar no **X**.

Parece pouco, mas é a **base de todo jogo**: todo o resto (personagem, carros, pontos) vai ser colocado dentro do que construirmos aqui.

Vamos dividir o bloco em 6 pedaços.

## 1.1 – Ligando o pygame

````python
import pygame

pygame.init()
````

### O que esse pedaço faz?
Antes de usar qualquer coisa do pygame, a gente precisa **trazer a biblioteca** para o programa e **ligar** ela.

🧠 **Analogia:** é como um videogame. Primeiro você **pega o console** da estante (`import`), depois **aperta o botão de ligar** (`init`). Sem ligar, nada funciona.

### Linha por linha
| Linha | O que faz |
|---|---|
| `import pygame` | Traz a biblioteca pygame para dentro do nosso programa. A partir daqui, podemos usar tudo que começa com `pygame.` |
| `pygame.init()` | "Liga" todas as partes do pygame: vídeo, teclado, som, etc. Sempre vem logo depois do import. |

### 💡 Detalhes
- Os **parênteses vazios** `()` em `init()` significam que estamos **chamando uma função** (mandando ela executar), mesmo sem passar nenhuma informação.
- O **ponto** em `pygame.init` quer dizer "a função `init` que mora **dentro** do pygame".

## 1.2 – Criando a janela

````python
LARGURA = 600
ALTURA = 600

tela = pygame.display.set_mode((LARGURA, ALTURA))
pygame.display.set_caption("Meu Crossy Road")
````

### O que esse pedaço faz?
Cria a **janela do jogo** com 600 pixels de largura por 600 de altura e coloca um título nela.

🧠 **Analogia:** a `tela` é como uma **folha de papel** em branco. Tudo que o jogo mostrar (fundo, personagem, carros) vai ser **desenhado nessa folha**.

### Linha por linha
| Linha | O que faz |
|---|---|
| `LARGURA = 600` | Guarda o número 600 numa variável chamada `LARGURA`. É a largura da janela, em pixels. |
| `ALTURA = 600` | Mesma coisa para a altura. |
| `tela = pygame.display.set_mode((LARGURA, ALTURA))` | **Cria a janela** com esse tamanho e guarda ela na variável `tela`. É nela que vamos desenhar. |
| `pygame.display.set_caption("Meu Crossy Road")` | Muda o **título** que aparece na barra de cima da janela. |

### 💡 Detalhes
- **Por que MAIÚSCULAS?** É uma **convenção** (combinado entre programadores): nome todo em maiúsculo indica uma **constante**, um valor que **não muda** durante o jogo. O Python não obriga, mas ajuda quem lê o código.
- **Por que dois parênteses `((...))`?** O de fora é da função `set_mode(...)`. O de dentro cria um **par de valores** `(600, 600)` — o pygame quer receber o tamanho como um "pacote" só, chamado **tupla**.
- **Pixel** é cada pontinho da tela. 600×600 = uma janela quadrada de tamanho médio.

### 📍 As coordenadas no pygame (MUITO importante)
```
(0,0) ──────────► x cresce para a DIREITA
  │
  │
  ▼
  y cresce para BAIXO         (600,600) fica no canto inferior direito
```
O ponto `(0, 0)` é o **canto de cima à esquerda**. Diferente da matemática, o **y aumenta para baixo**. Isso vai importar muito quando o personagem andar!

## 1.3 – O relógio e as cores

````python
relogio = pygame.time.Clock()

VERDE = (100, 200, 100)
````

### O que esse pedaço faz?
Cria um **relógio** para controlar a velocidade do jogo e define a **cor verde** que vamos usar no fundo.

### Linha por linha
| Linha | O que faz |
|---|---|
| `relogio = pygame.time.Clock()` | Cria um relógio e guarda na variável `relogio`. Vamos usar ele lá no final do loop para o jogo não rodar rápido demais. |
| `VERDE = (100, 200, 100)` | Guarda uma cor. Toda cor no pygame são **3 números de 0 a 255**: (Vermelho, Verde, Azul). |

### 🎨 Como funcionam as cores (RGB)
Cada número diz **quanta luz** de cada cor tem. 0 = nada, 255 = o máximo.

| Cor | Valor |
|---|---|
| Preto | `(0, 0, 0)` — nenhuma luz |
| Branco | `(255, 255, 255)` — todas no máximo |
| Vermelho | `(255, 0, 0)` |
| Azul | `(0, 0, 255)` |
| Nosso verde | `(100, 200, 100)` — bastante verde, um pouco de vermelho e azul (fica um verde-grama clarinho) |

## 1.4 – O loop principal e os eventos

````python
rodando = True
while rodando:

    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
````

### O que esse pedaço faz?
Esse é o **coração do jogo**. Um jogo é basicamente um `while` que se repete **dezenas de vezes por segundo**, sem parar, até o jogador sair.

Aqui dentro, a primeira coisa que fazemos em cada volta é **ver o que o jogador fez** (apertou tecla? clicou no X?).

🧠 **Analogia:** é como um **desenho animado em flipbook**. Cada volta do `while` desenha **uma página** (um *quadro* ou *frame*). Passando muitas páginas por segundo, parece que as coisas se mexem.

### Linha por linha
| Linha | O que faz |
|---|---|
| `rodando = True` | Cria uma variável que diz se o jogo **está ligado**. Começa como `True` (verdadeiro). |
| `while rodando:` | "Enquanto `rodando` for verdadeiro, repita tudo que está aqui dentro." |
| `for evento in pygame.event.get():` | Pega a **lista de tudo que aconteceu** desde a última volta (teclas, cliques, fechar janela) e olha **um por um**. |
| `if evento.type == pygame.QUIT:` | Pergunta: "esse evento foi o jogador clicando no **X** da janela?" |
| `rodando = False` | Se foi, muda `rodando` para falso. Na próxima vez que o `while` checar, ele **para**. |

### 💡 Detalhes
- **Evento** = qualquer coisa que o jogador faz: apertar tecla, soltar tecla, mexer o mouse, clicar no X.
- `pygame.QUIT` é o **nome do evento** "fechar a janela".
- Repare na **indentação** (espaços no começo da linha): é ela que diz o que está **dentro** do `while`, do `for` e do `if`.
- Por que não usar `break`? Com a variável `rodando`, o loop termina a volta atual inteira e sai de forma "limpa". É mais fácil de ler.

## 1.5 – Desenhar, mostrar e esperar
*(este pedaço continua **dentro** do `while`, por isso tem espaço na frente)*

````python
    tela.fill(VERDE)

    pygame.display.update()

    relogio.tick(60)
````

### O que esse pedaço faz?
Em toda volta do loop: **pinta** a tela de verde, **mostra** o resultado na janela e **espera** um pouquinho para manter a velocidade certa.

### Linha por linha
| Linha | O que faz |
|---|---|
| `tela.fill(VERDE)` | Pinta a **tela inteira** de verde. Fazemos isso todo quadro para **apagar** o desenho anterior (senão, quando o personagem andar, ele deixaria um "rastro"). |
| `pygame.display.update()` | **Mostra** na janela tudo que foi desenhado. Antes disso, o desenho fica "escondido". |
| `relogio.tick(60)` | Faz o loop rodar **no máximo 60 vezes por segundo** (60 FPS). |

### 💡 Detalhes
- **Por que desenhar "escondido" e depois mostrar?** Se cada coisa aparecesse na hora em que é desenhada, a tela ficaria piscando. Então o pygame desenha tudo primeiro e mostra **de uma vez só** no `update()`.
- **FPS** = *frames per second*, quadros por segundo. 60 é o padrão da maioria dos jogos.
- **Sem o `tick`**, o loop rodaria o mais rápido que o computador conseguisse — e em cada computador o jogo teria uma velocidade diferente!

### 🔁 Resumindo a ordem de cada volta do loop
1. **Ler eventos** — o que o jogador fez? *(1.4)*
2. **Atualizar** — mexer personagem, carros, checar batidas *(vamos fazer nos próximos blocos)*
3. **Desenhar** — `fill` e depois os desenhos
4. **Mostrar e esperar** — `update` e `tick`

## 1.6 – Desligando o pygame
*(este pedaço fica **fora** do `while`, sem espaço na frente)*

````python
pygame.quit()
````

### O que esse pedaço faz?
Quando o `while` termina (o jogador clicou no X), o programa continua para a linha de baixo, que **desliga o pygame** e fecha a janela.

### Linha por linha
| Linha | O que faz |
|---|---|
| `pygame.quit()` | Desliga o pygame (o contrário do `pygame.init()`). A janela fecha. |

🧠 **Analogia:** é o botão de **desligar** o videogame antes de guardar.

```

    fill              →   pinta o fundo
```

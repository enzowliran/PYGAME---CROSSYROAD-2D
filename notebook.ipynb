# 🧱 BLOCO 1 – Janela e loop principal
**Objetivo:** abrir uma janela verde que fica aberta até clicar no X.

```python
import pygame

pygame.init()
```

**Liga o pygame.**
| Linha | O que faz |
|---|---|
| `import pygame` | Traz a biblioteca para o programa |
| `pygame.init()` | Liga as partes do pygame (vídeo, teclado, som). Sempre logo após o import |

```python
LARGURA = 600
ALTURA = 600

tela = pygame.display.set_mode((LARGURA, ALTURA))
pygame.display.set_caption("Segura Aí!")
```

**Cria a janela.** A `tela` é onde tudo vai ser desenhado.
| Linha | O que faz |
|---|---|
| `LARGURA = 600` / `ALTURA = 600` | Tamanho da janela em pixels. MAIÚSCULO = constante (valor que não muda) |
| `set_mode((LARGURA, ALTURA))` | Cria a janela e guarda em `tela`. Parênteses duplos: o tamanho vai como um par `(600, 600)` |
| `set_caption(...)` | Título da janela |

📍 **Coordenadas:** `(0, 0)` é o **canto superior esquerdo**. x cresce para a direita, **y cresce para baixo**.

```python
relogio = pygame.time.Clock()

VERDE = (100, 200, 100)
```

| Linha | O que faz |
|---|---|
| `relogio = pygame.time.Clock()` | Relógio para controlar a velocidade do jogo (usado no `tick`) |
| `VERDE = (100, 200, 100)` | Cor em RGB: (Vermelho, Verde, Azul), cada um de 0 a 255. Ex: `(0,0,0)` preto, `(255,255,255)` branco |

```python
rodando = True
while rodando:

    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
```

**Loop principal:** o jogo é um `while` que repete ~60 vezes por segundo. Cada volta = 1 quadro (frame).
| Linha | O que faz |
|---|---|
| `rodando = True` | Controla se o jogo continua |
| `while rodando:` | Repete enquanto `rodando` for `True` |
| `for evento in pygame.event.get():` | Percorre a lista do que o jogador fez (teclas, cliques, X) |
| `if evento.type == pygame.QUIT:` | O jogador clicou no X? |
| `rodando = False` | Então o `while` para |

```python
    tela.fill(VERDE)

    pygame.display.update()

    relogio.tick(60)
```

*(continua dentro do `while`)*
| Linha | O que faz |
|---|---|
| `tela.fill(VERDE)` | Pinta a tela toda. Feito todo quadro para **apagar** o anterior (senão fica rastro) |
| `pygame.display.update()` | **Mostra** o que foi desenhado. Sem ele, a janela fica preta |
| `relogio.tick(60)` | Limita a 60 FPS, para ter a mesma velocidade em qualquer PC |

🔁 **Ordem de todo quadro:** eventos → atualizar → desenhar → `update` → `tick`

```python
pygame.quit()
```

*(fora do `while`)* Desliga o pygame e fecha a janela quando o loop acaba.

# 🧱 BLOCO 2 – Personagem andando com as setas
**Objetivo:** um quadrado azul (o personagem) que anda **uma casa por vez** com as setas.

🗂️ **Como organizamos o jogo (sem classes):** os **dados** ficam em **dicionários** e as **ações** em **funções**.

```python
AZUL = (50, 100, 255)

CASA = 50

personagem = {"x": 250, "y": 550}
```

| Linha | O que faz |
|---|---|
| `AZUL = (50, 100, 255)` | Cor do personagem |
| `CASA = 50` | Tamanho de cada "casa" do chão. O personagem anda 50 pixels por vez (como num tabuleiro). A tela 600×600 vira 12×12 casas |
| `personagem = {"x": 250, "y": 550}` | O personagem é um **dicionário** com sua posição. `x=250` ≈ meio da tela, `y=550` = última linha (lá embaixo) |

💡 `x` e `y` são o **canto superior esquerdo** do quadrado.

```python
def mover_personagem(personagem, tecla):
    if tecla == pygame.K_UP and personagem["y"] > 0:
        personagem["y"] = personagem["y"] - CASA
    if tecla == pygame.K_DOWN and personagem["y"] < ALTURA - CASA:
        personagem["y"] = personagem["y"] + CASA
    if tecla == pygame.K_LEFT and personagem["x"] > 0:
        personagem["x"] = personagem["x"] - CASA
    if tecla == pygame.K_RIGHT and personagem["x"] < LARGURA - CASA:
        personagem["x"] = personagem["x"] + CASA
```

**Recebe a tecla apertada e anda uma casa naquela direção.**
| Linha | O que faz |
|---|---|
| `def mover_personagem(personagem, tecla):` | Cria a função. Recebe o dicionário do personagem e a tecla |
| `if tecla == pygame.K_UP and personagem["y"] > 0:` | Apertou ↑ **e** não está no topo? |
| `personagem["y"] = personagem["y"] - CASA` | Sobe: **diminui** o y (lembre: y cresce para baixo) |
| ↓ | Aumenta o y, se não estiver na última linha (`ALTURA - CASA` = 550) |
| ← / → | Diminui / aumenta o x, com o mesmo cuidado nas bordas |

💡 **Por que `ALTURA - CASA` e não `ALTURA`?** Como x/y é o canto de cima do quadrado, se y fosse 600 ele estaria **fora** da tela. A última posição possível é 600 − 50 = 550.

💡 **A função muda o dicionário original.** Quando passamos um dicionário para uma função e alteramos ele lá dentro, a mudança vale fora também. Por isso não precisa de `return`.

```python
def desenhar_personagem(tela, personagem):
    pygame.draw.rect(tela, AZUL, (personagem["x"], personagem["y"], CASA, CASA))
```

| Linha | O que faz |
|---|---|
| `pygame.draw.rect(tela, cor, (x, y, largura, altura))` | Desenha um retângulo na `tela`. Aqui: na posição do personagem, com 50×50 |

```python
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            rodando = False
        if evento.type == pygame.KEYDOWN:
            mover_personagem(personagem, evento.key)

    tela.fill(VERDE)
    desenhar_personagem(tela, personagem)
```

**Mudanças dentro do loop principal:**
| Linha | O que faz |
|---|---|
| `if evento.type == pygame.KEYDOWN:` | Alguma tecla foi **apertada**? |
| `mover_personagem(personagem, evento.key)` | Chama a função passando **qual** tecla foi (`evento.key`) |
| `desenhar_personagem(tela, personagem)` | Desenha o personagem. Vem **depois** do `fill`, senão o fundo cobriria ele |

💡 **KEYDOWN** acontece **uma vez** por apertada. Por isso o personagem anda uma casa, e não sai deslizando enquanto a tecla está segurada.

3. Por que `mover_personagem` não tem `return`? <details><summary>R</summary>Ela altera o dicionário recebido, e isso muda o original.</details>

# 🧱 BLOCO 3 – Obstáculos andando no corredor
**Objetivo:** pessoas e carrinhos que andam sozinhos pelo corredor, de um lado para o outro.

💡 Todo obstáculo é **o mesmo tipo de dicionário**; só mudam os valores (velocidade, largura, cor).

```python
LARANJA = (255, 150, 50)
VERMELHO = (220, 50, 50)

def criar_obstaculo(x, y, velocidade, largura, cor):
    return {"x": x, "y": y, "velocidade": velocidade, "largura": largura, "cor": cor}
```

**Fábrica de obstáculos:** recebe os valores e **devolve** (`return`) um dicionário pronto.
| Linha | O que faz |
|---|---|
| `LARANJA` / `VERMELHO` | Cores novas: aluno (laranja) e carrinho de limpeza (vermelho) |
| `def criar_obstaculo(...)` | Função com 5 parâmetros |
| `return {...}` | Monta o dicionário com os valores recebidos e devolve |

💡 Assim não precisamos escrever o dicionário inteiro toda vez.

```python
obstaculos = [
    criar_obstaculo(0, 450, 2, 50, LARANJA),
    criar_obstaculo(300, 450, 2, 50, LARANJA),
    criar_obstaculo(100, 350, -5, 100, VERMELHO),
    criar_obstaculo(400, 250, 3, 50, LARANJA),
    criar_obstaculo(200, 150, -6, 100, VERMELHO),
]
```

**Lista com todos os obstáculos** (uma lista de dicionários).
| Valor | Significado |
|---|---|
| `y` (450, 350...) | Em qual faixa ele anda. Sempre múltiplo de 50 para encaixar nas casas |
| `velocidade` | Pixels por quadro. **Positivo → direita**, **negativo → esquerda** |
| `largura` | 50 = aluno (1 casa), 100 = carrinho (2 casas) |

Ex: `criar_obstaculo(100, 350, -5, 100, VERMELHO)` = carrinho na faixa y=350, rápido, indo para a esquerda.

```python
def mover_obstaculos(obstaculos):
    for obstaculo in obstaculos:
        obstaculo["x"] = obstaculo["x"] + obstaculo["velocidade"]
        if obstaculo["x"] > LARGURA:
            obstaculo["x"] = -obstaculo["largura"]
        if obstaculo["x"] < -obstaculo["largura"]:
            obstaculo["x"] = LARGURA
```

**Anda com todos e faz quem sai da tela voltar pelo outro lado.**
| Linha | O que faz |
|---|---|
| `for obstaculo in obstaculos:` | Passa por cada dicionário da lista |
| `x = x + velocidade` | Anda. Se a velocidade é negativa, o x diminui (vai para a esquerda) |
| `if x > LARGURA:` → `x = -largura` | Saiu pela **direita**: reaparece escondido logo antes da borda **esquerda** |
| `if x < -largura:` → `x = LARGURA` | Saiu pela **esquerda** (sumiu por inteiro): reaparece na borda **direita** |

💡 Usamos `-largura` para ele entrar na tela **aos poucos**, e não "piscar" já dentro dela.

```python
def desenhar_obstaculos(tela, obstaculos):
    for obstaculo in obstaculos:
        pygame.draw.rect(tela, obstaculo["cor"], (obstaculo["x"], obstaculo["y"], obstaculo["largura"], CASA))
```

Igual ao `desenhar_personagem`, mas para cada obstáculo, usando a **cor** e a **largura** dele. A altura é sempre `CASA` (50).

```python
    mover_obstaculos(obstaculos)

    tela.fill(VERDE)
    desenhar_obstaculos(tela, obstaculos)
    desenhar_personagem(tela, personagem)
```

**Mudanças dentro do loop principal:**
| Linha | O que faz |
|---|---|
| `mover_obstaculos(obstaculos)` | Fica **fora** do `for evento`: os obstáculos andam em **todo quadro**, mesmo sem apertar tecla |
| `desenhar_obstaculos(...)` | Depois do `fill` e **antes** do personagem, para o personagem aparecer por cima |

🔁 Agora o loop tem as 3 etapas: **eventos → atualizar → desenhar**.
